```python
#importing the Libraies
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset=pd.read_csv("Social_Network_Ads.csv")
```

```python
dataset
```

```python
dataset=pd.get_dummies(dataset,drop_first=True, dtype=int)
dataset
```

```python
dataset.columns
dataset=dataset.drop("User ID",axis=1)
```

```python
dataset["Purchased"].value_counts()
```

```python
indep=dataset[["Age","EstimatedSalary","Gender_Male"]]
dep=dataset["Purchased"]
```

```python
indep.shape
```

```python
dep
```

```python
#split into training set and test
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(indep, dep, test_size = 1/3, random_state = 0)
```

```python
from sklearn.naive_bayes import MultinomialNB
classifier = MultinomialNB()
classifier.fit(X_train, y_train)
y_pred = classifier.predict(X_test)
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
from sklearn.metrics import classification_report
clf_report = classification_report(y_test, y_pred)              
print(clf_report)
print(cm)

# 1. waht is the recall of (0)? ans => 1.00
#(or) what is the correct performance of non purches? ans => 1.00

# 2. waht is the recall of (1)? ans => 0.00 
#(or) what is the correct performance of  purches? ans => 0.00 

#3.what is the over all performance of accuracy?  ans => 0.63

#4.what is the performance of  precision (0) or non purches?  ans => 0.63
#5.what is the performance of  precision (1) or  purches?  ans => 0.00

#6.what is the performance of f1-score of (0)?  ans => 0.78
#7.what is the performance of f1-score of (1)?  ans => 0.00

#3.what is the macro avg of precision?  ans => 0.32
#3.what is the macro avg of recall?  ans => 0.50
#3.what is the macro avg of  f1-score?  ans => 0.39
#3.what is the weighted avg of precision?  ans => 0.40
#3.what is the weighted avg of recall?  ans => 0.63
#3.what is the weighted avg of f1-score?  ans => 0.43
```
